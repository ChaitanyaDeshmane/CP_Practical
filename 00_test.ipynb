```python
#If elif else statement for marks of the students
marks = int(input("Enter the marks of the student: "))

if marks >= 95:
    grade = "A"
elif marks >= 85:
    grade = "B"
elif marks >= 80:
    grade = "C"
elif marks >= 75:
    grade = "D"
else:
    grade = "F"

print(f"The grade of the student is: {grade}")
```
